```python
import numpy as np
import pandas as pd
```

```python
# Example population data
data = pd.DataFrame({
    "ID": range(1, 101),
    "Gender": np.random.choice(["Male", "Female"], 100),
    "Score": np.random.randint(40, 100, 100)
})
```

```python

# Create clusters (10 clusters)
data["Cluster"] = data["ID"] // 10

```

```python
selected_clusters = np.random.choice(data["Cluster"].unique(), size=2, replace=False)

sample_cluster = data[data["Cluster"].isin(selected_clusters)]
print(sample_cluster)
```
