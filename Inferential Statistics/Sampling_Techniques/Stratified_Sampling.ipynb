```python
import numpy as np
import pandas as pd
```

# data creation

```python
# Example population data
data = pd.DataFrame({
    "ID": range(1, 101),
    "Gender": np.random.choice(["Male", "Female"], 100),
    "Score": np.random.randint(40, 100, 100)
})
```

# stratified sampling

```python
stratifies_sample=data.groupby("Gender").apply(lambda x : x.sample(n=5,random_state=42))
stratifies_sample
```
