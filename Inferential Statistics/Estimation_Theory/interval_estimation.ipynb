```python
import numpy as np
from scipy import stats
```

```python
data = [10, 12, 14, 15, 13]
```

```python
mean=np.mean(data)
std=np.std(data)
n=len(data)
```

# z-score

```python
confidence_value=0.95
z=stats.norm.ppf((1+confidence_value)/2)
z
```

# margin of error

```python
error=z*(std/np.sqrt(n))
```

# confidence interval

```python
ci=(mean-error,mean+error)
ci
```
