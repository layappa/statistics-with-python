✅ Complete Hypothesis Testing Steps
Step-by-step Process:

State H₀ and H₁

Choose α

Compute test statistic

Find critical value

Compare statistic with critical value

Make decision

✅ Example with Formula + Python Code
Problem:

A company claims mean weight is 500g.
Sample of 30 packets has:

x̄ = 492

σ = 20

α = 0.05

```python
import numpy as np
from scipy import stats

# Given data
x_bar = 492
mu = 500
sigma = 20
n = 30
alpha = 0.05

# Compute Z statistic
z = (x_bar - mu) / (sigma / np.sqrt(n))
print("Z Statistic:", z)

# Critical value for two-tailed test
critical = stats.norm.ppf(1 - alpha/2)
print("Critical Value:", critical)

# Decision
if abs(z) > critical:
    print("Reject H0 (Significant difference)")
else:
    print("Fail to Reject H0 (No significant difference)")
```
