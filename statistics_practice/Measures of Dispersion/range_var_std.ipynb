# *range*


```python
data = [10, 20, 30, 40, 50]
range_val = max(data) - min(data)
print(range_val)
```

# *VARIANCE*


```python
import numpy as np
data = [10, 20, 30, 40, 50]
variance = np.var(data)
print(variance)
```

# *standard deviation*

```python
import numpy as np
data = [10, 20, 30, 40, 50]
std_dev = np.std(data)
print(std_dev)
```

# *IQR*


```python
import numpy as np

data = [10, 20, 30, 40, 50, 60, 70]

Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

iqr = Q3 - Q1
print(iqr)
```
