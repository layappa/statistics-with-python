```python
# Empirical Rule (68-95-99.7 Rule)

def empirical_rule(mean, std):
    ranges = {
        "68_percent": (mean - std, mean + std),
        "95_percent": (mean - 2*std, mean + 2*std),
        "99_7_percent": (mean - 3*std, mean + 3*std)
    }
    return ranges




```

```python
# Example — Student Heights

mean = 160   # μ
std  = 10    # σ

result = empirical_rule(mean, std)
```

```python
print("Empirical Rule Ranges for Student Heights\n")

print(f"68% of students: {result['68_percent'][0]} to {result['68_percent'][1]} cm")
print(f"95% of students: {result['95_percent'][0]} to {result['95_percent'][1]} cm")
print(f"99.7% of students: {result['99_7_percent'][0]} to {result['99_7_percent'][1]} cm")
```
