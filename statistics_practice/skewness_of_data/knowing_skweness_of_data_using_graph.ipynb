```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import skew, kurtosis

# -----------------------------
# Generate data
# -----------------------------
np.random.seed(1)

uniform_data   = np.random.uniform(0, 100, 1000)
normal_data    = np.random.normal(50, 10, 1000)      # symmetric
right_skewed   = np.random.exponential(10, 1000)    # right skewed
left_skewed    = -np.random.exponential(10, 1000)   # left skewed

# Kurtosis examples
peaked_data = np.random.laplace(0, 1, 1000)   # leptokurtic (peaked)
flat_data   = np.random.uniform(-5, 5, 1000) # platykurtic (flat)

# -----------------------------
# Plot function
# -----------------------------
def plot_dist(data, title):
    plt.hist(data, bins=40)
    plt.title(title)
    plt.xlabel("Values")
    plt.ylabel("Frequency")
    plt.show()
    
    print(title)
    print("Skewness:", skew(data))
    print("Kurtosis:", kurtosis(data))
    print("-"*40)

# -----------------------------
# Draw graphs
# -----------------------------
plot_dist(uniform_data, "Uniform Distribution")
plot_dist(normal_data, "Symmetric (Normal) Distribution")
plot_dist(right_skewed, "Right Skewed Distribution")
plot_dist(left_skewed, "Left Skewed Distribution")
plot_dist(peaked_data, "Peaked (Leptokurtic) Distribution")
plot_dist(flat_data, "Flat (Platykurtic) Distribution")
```
