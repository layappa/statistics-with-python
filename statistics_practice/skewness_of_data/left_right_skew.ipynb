```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import skew, kurtosis

# Sample Data (you can change this)
data = np.array([12, 15, 18, 20, 22, 23, 24, 25, 26, 28, 30])

# -------------------------------
# 1️⃣ Skewness
# -------------------------------
sk = skew(data)

# -------------------------------
# 2️⃣ Kurtosis
# -------------------------------
kt = kurtosis(data)   # Fisher’s definition (0 = normal)

# -------------------------------
# 3️⃣ Distribution Type
# -------------------------------
if abs(sk) < 0.5:
    skew_type = "Symmetric"
elif sk > 0:
    skew_type = "Right Skewed (Positive Skew)"
else:
    skew_type = "Left Skewed (Negative Skew)"

# -------------------------------
# 4️⃣ Kurtosis Type
# -------------------------------
if kt > 0:
    kurt_type = "Leptokurtic (Peaked)"
elif kt < 0:
    kurt_type = "Platykurtic (Flat)"
else:
    kurt_type = "Mesokurtic (Normal)"

# -------------------------------
# 5️⃣ Check Uniform Distribution
# -------------------------------
# If data values are evenly spread
hist, _ = np.histogram(data, bins=5)
if np.std(hist) < 2:
    uniform = "Uniform Distribution"
else:
    uniform = "Not Uniform"

# -------------------------------
# 6️⃣ Print Results
# -------------------------------
print("Skewness:", sk)
print("Skew Type:", skew_type)
print("Kurtosis:", kt)
print("Kurtosis Type:", kurt_type)
print("Uniformity:", uniform)

# -------------------------------
# 7️⃣ Plot Histogram
# -------------------------------
plt.hist(data, bins=10)
plt.title("Data Distribution")
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.show()
```
